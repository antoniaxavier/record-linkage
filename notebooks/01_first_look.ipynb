## Amazon

```python
# getting data
amazon = pl.read_csv("../data/raw/Amazon.csv", encoding="latin1")
```

```python
# dando uma olhada
amazon.glimpse()
```

colunas: 
id: codigo identificador da linha
title: titulo do anuncio
description: descricao do anuncio 
manufacturer: fabricante
price: preço em usd

```python
amazon.schema                         # nomes e tipos das colunas (≈ str())
```

```python

amazon.null_count()                   # NAs por coluna (≈ colSums(is.na(df)))
```

```python
amazon["manufacturer"].value_counts(sort=True)   # ≈ count(df, manufacturer, sort = TRUE)
```

```python
amazon["manufacturer"].n_unique()     # ≈ n_distinct()
```
